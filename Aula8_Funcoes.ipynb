## 📌 1. O que é uma Função?


Uma função é um **bloco de código reutilizável** que executa uma tarefa específica. Em vez de escrever o mesmo código várias vezes, colocamos esse código dentro de uma função e chamamos essa função quando necessário.

## 🔁 5. Funções com Retorno

````python
def somar(a, b):
    return a + b

resultado = somar(5, 3)
print(f"A soma é: {resultado}")
````

## ❓ 6. Diferença entre `print` e `return`

````python
def exemplo_print(x):
    print(x * 2)

def exemplo_return(x):
    return x * 2

exemplo_print(5)
resultado = exemplo_return(5)
print(resultado)
````

## 📚 7. Exemplos Aplicados

````python
def eh_par(numero):
    return numero % 2 == 0

print(eh_par(4))
print(eh_par(7))
````

````python
def fatorial(n):
    resultado = 1
    for i in range(2, n + 1):
        resultado *= i
    return resultado

print(fatorial(5))
````

````python
def calcular_media(n1, n2):
    return (n1 + n2) / 2

print(calcular_media(7, 9))
````

4. Crie uma função chamada eh_primo(numero) que retorne True se o número for primo.

````python
numero = int(input("Digite um número: "))

def eh_primo(numero):
  if numero < 2:
    return False
  for i in range(2, numero):
    if numero % i == 0:
      return False
  return True

eh_primo(numero)
````

11. Desafio Final (Avançado)
Simule um sistema de cadastro de produtos com as seguintes opções:

1 - Cadastrar produto  
2 - Listar produtos  
3 - Buscar produto pelo nome  
4 - Sair
Use uma lista para armazenar os produtos.
Crie funções para cada opção.
Utilize laços de repetição e estruturas de decisão junto com funções.

````python
produtos = []
````

````python
def cadastrar_produto():
    nome = input("Digite o nome do produto: ")
    while True:
        try:
            preco = float(input("Digite o preço do produto: "))
            break
        except ValueError:
            print("Preço inválido. Por favor, digite um número.")
    produto = {"nome": nome, "preco": preco}
    produtos.append(produto)
    print(f"Produto '{nome}' cadastrado com sucesso!")
````

````python
def listar_produtos():
    if not produtos:
        print("Nenhum produto cadastrado.")
        return
    print("\n--- Lista de Produtos ---")
    for i, produto in enumerate(produtos):
        print(f"{i+1}. Nome: {produto['nome']}, Preço: R${produto['preco']:.2f}")
    print("-------------------------")
````

````python
def buscar_produto_por_nome():
    if not produtos:
        print("Nenhum produto cadastrado para buscar.")
        return
    nome_busca = input("Digite o nome do produto para buscar: ").lower()
    encontrados = [p for p in produtos if nome_busca in p['nome'].lower()]
    if encontrados:
        print("\n--- Produtos Encontrados ---")
        for produto in encontrados:
            print(f"Nome: {produto['nome']}, Preço: R${produto['preco']:.2f}")
        print("----------------------------")
    else:
        print(f"Nenhum produto encontrado com o nome '{nome_busca}'.")
````

````python
def exibir_menu():
    print("\n--- Menu Principal ---")
    print("1 - Cadastrar produto")
    print("2 - Listar produtos")
    print("3 - Buscar produto pelo nome")
    print("4 - Sair")
    print("----------------------")
````

````python
def main():
    while True:
        exibir_menu()
        opcao = input("Escolha uma opção: ")

        if opcao == '1':
            cadastrar_produto()
        elif opcao == '2':
            listar_produtos()
        elif opcao == '3':
            buscar_produto_por_nome()
        elif opcao == '4':
            print("Saindo do programa. Até mais!")
            break
        else:
            print("Opção inválida. Por favor, tente novamente.")

# Chamada da função principal para iniciar o sistema
main()
````
