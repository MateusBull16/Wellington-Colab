## 📌 1. O que é uma Função?


Uma função é um **bloco de código reutilizável** que executa uma tarefa específica. Em vez de escrever o mesmo código várias vezes, colocamos esse código dentro de uma função e chamamos essa função quando necessário.

## 🧪 3. Funções Sem Parâmetros e Sem Retorno

````python
def exibir_mensagem():
    print("Olá! Esta é uma função sem parâmetros e sem retorno.")

exibir_mensagem()
````

## ✉️ 4. Funções com Parâmetros

````python
def saudacao(nome):
    print(f"Olá, {nome}! Seja bem-vindo.")

saudacao("Carlos")
saudacao("Marina")
````

## 🔁 5. Funções com Retorno

````python
def somar(a, b):
    return a + b

resultado = somar(5, 3)
print(f"A soma é: {resultado}")
````

## ❓ 6. Diferença entre `print` e `return`

````python
def exemplo_print(x):
    print(x * 2)

def exemplo_return(x):
    return x * 2

exemplo_print(5)
resultado = exemplo_return(5)
print(resultado)
````

## 📚 7. Exemplos Aplicados

````python
def eh_par(numero):
    return numero % 2 == 0

print(eh_par(4))
print(eh_par(7))
````

````python
def fatorial(n):
    resultado = 1
    for i in range(2, n + 1):
        resultado *= i
    return resultado

print(fatorial(5))
````

````python
def calcular_media(n1, n2):
    return (n1 + n2) / 2

print(calcular_media(7, 9))
````

4. Crie uma função chamada eh_primo(numero) que retorne True se o número for primo.

````python
numero = int(input("Digite um número: "))

def eh_primo(numero):
  if numero < 2:
    return False
  for i in range(2, numero):
    if numero % i == 0:
      return False
  return True

eh_primo(numero)
````
